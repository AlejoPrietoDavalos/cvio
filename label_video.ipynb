```python
from typing import List, Tuple, Optional
from pathlib import Path
from PIL import Image, ImageTk

import tkinter as tk
import cv2


def get_points_from_image(
        *,
        img: Image.Image,
        n_points: int,
        color: str = "yellow"
) -> List[Tuple[int, int]]:
    points: List[Tuple[int, int]] = []

    def on_click(event: tk.Event) -> None:
        """ Pone un punto cuando se da click, y agrega las coordenadas a `points`."""
        if len(points) < n_points:
            x, y = event.x, event.y
            points.append((x, y))
            canvas.create_oval(x-3, y-3, x+3, y+3, fill=color, outline=color)
            if len(points) == n_points:
                root.after(500, root.destroy)

    root = tk.Tk()
    root.title("-> Hacé click para agregar puntos")

    tk_image = ImageTk.PhotoImage(img)
    canvas = tk.Canvas(root, width=img.width, height=img.height)
    canvas.pack()
    canvas.create_image(0, 0, anchor=tk.NW, image=tk_image)
    canvas.bind("<Button-1>", on_click)

    root.mainloop()
    return points

```

```python
def get_first_frame(path_video: Path) -> Optional[Image.Image]:
    cap = cv2.VideoCapture(str(path_video))
    success, frame = cap.read()
    cap.release()

    if not success:
        return None

    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    return Image.fromarray(frame_rgb)

```

```python
path_video = Path("data") / "video_example.mp4"
img = get_first_frame(path_video)
img
```

```python
points_ref = get_points_from_image(img=img, n_points=9)
points_ref
```

```python
import json

with open(path_video.with_suffix(".json"), "w") as f:
    json.dump(points_ref, f)
```
